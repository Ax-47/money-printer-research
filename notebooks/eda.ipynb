```python
from pathlib import Path

import kagglehub
import polars as pl

pl.Config.set_tbl_cols(-1)  # แสดงทุกคอลัมน์ ไม่ซ่อนตรงกลาง
pl.Config.set_tbl_rows(6)

files = {}
for f in sorted(Path("../data").rglob("*.csv")):
    if f.parent.name == "lineups_csv" and f.name != "lineups_all.csv":
        continue
    files[str(f)] = pl.read_csv(f, infer_schema_length=10000, try_parse_dates=True)

kg_dir = Path(
    kagglehub.dataset_download("marcohuiii/english-premier-league-epl-match-data-2000-2025")
)
files["kaggle/epl_final.csv"] = pl.read_csv(kg_dir / "epl_final.csv", try_parse_dates=True)

for name, df in files.items():
    print("=" * 80)
    print(name, "| rows:", df.height, "| cols:", df.width)
    nulls = {c: n for c, n in zip(df.columns, df.null_count().row(0)) if n}
    print("nulls:", nulls or "none")
    display(df.head(5))
```
